Este cuaderno ejecuta la simulación (`modulacion.py`) y permite **ver** las figuras y **escuchar** las señales.

Sistema de 2 canales multiplexados en frecuencia (FDM):
- **Canal 1:** tono de prueba $m_1(t)=50\cos(2\pi\,440\,t)$ V
- **Canal 2:** 5 s de *La Marcha Imperial* (mono, limitado a 5 kHz, $|m_2|_{max}=50$ V)

Esquemas: AM (detección de envolvente), FM banda angosta ($\beta=0{,}2$), FM banda ancha ($\beta=5$) y PM ($k_p=0{,}1$ rad/V).

```
import json
from pathlib import Path
from IPython.display import Audio, Image, display, Markdown
import modulacion as mod

res = mod.main()   # genera assets/figs, assets/audio y assets/resultados.json (~20 s)
```

## 1. Tono puro en AM

```
t = res['tono']['AM']
print(f"a = {res['tono']['a']}   P_T = {t['PT_med']:.1f} W   Pc = {t['Pc_med']:.1f} W   P_B = {t['PB_med']:.1f} W   E% = {t['E_med']:.2f} %   B_T = {t['BT']:.0f} Hz")
for f in ['am_tono_tiempo', 'am_tono_espectro']:
    display(Image(mod.FIGS / f'{f}.png'))
```

## 2. Tono puro en FM y PM

```
for k in ['FM_BA', 'FM_BW', 'PM']:
    t = res['tono'][k]
    print(f"{k:6s} β = {t['beta']:<4}  Δf = {t['df']:.0f} Hz   P_T = {t['PT_med']:.0f} W   Pc = {t['Pc_med']:.1f} W   B_T(Carson) = {t['BT_carson']:.0f} Hz")
for f in ['angular_tono_espectros']:
    display(Image(mod.FIGS / f'{f}.png'))
```

## 3. Sistema multicanal (FDM) con la Marcha Imperial

```
display(Image(mod.FIGS / 'audio_mensaje.png'))
display(Image(mod.FIGS / 'fdm_espectros.png'))
display(Image(mod.FIGS / 'fdm_audio_demodulado.png'))
for k, v in res['fdm'].items():
    for canal in ['tono', 'audio']:
        a = v[canal]
        print(f"{k:6s} {canal:5s} B_T = {a['BT']/1e3:.1f} kHz  Si/Ni = {a['SiNi_dB']:.1f} dB  So/No = {a['SoNo_dB']:.1f} dB (teo {a['SoNo_teo_dB']:.1f})  fidelidad = {a['fid_con_ruido_dB']:.1f} dB")
```
